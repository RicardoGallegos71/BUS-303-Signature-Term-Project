```python
import pandas as pd
import matplotlib.pyplot as plt

sbux4 = pd.read_csv('sbux4.csv')
dnkn4 = pd.read_csv('dnkn_clean.csv')

print("Starbucks:", sbux4.shape)
print("Dunkin:", dnkn4.shape)
```

# Task 3.1: Compare beverage portfolio sizes

```python
portfolio = pd.DataFrame({
    'Company': ['Starbucks', "Dunkin' Donuts"],
    'Beverage Combinations': [len(sbux4), len(dnkn4)]
})

starbucks_green = '#00704A'
dunkin_pink = '#FF59E9'

fig, ax = plt.subplots(figsize=(9, 4))

bars = ax.barh(
    portfolio['Company'],
    portfolio['Beverage Combinations'],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, padding=5, fontweight='bold')

ax.set_title('Beverage Combination Portfolio Size',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Number of Beverage Combinations')
ax.set_ylabel('Company')
ax.set_xlim(0, 1400)
ax.invert_yaxis()

plt.tight_layout()
plt.show()
```

This chart compares the number of beverage combinations offered by Starbucks and Dunkin' Donuts. Starbucks has 1,203 beverage combinations, while Dunkin' Donuts has 599. This shows that Starbucks has a larger beverage portfolio, with about twice as many combinations as Dunkin'. However, having more combinations does not necessarily mean that Starbucks offers healthier options.

## Task 3.2: Average Beverage Calories

This chart compares the average calories per beverage at Starbucks and Dunkin' Donuts.

```python
print("Starbucks columns:", sbux4.columns.tolist())
print("Dunkin columns:", dnkn4.columns.tolist())
```

```python
# Calculate average calories for each company
avg_calories = pd.DataFrame({
    'Company': ['Starbucks', "Dunkin' Donuts"],
    'Average_Calories': [
        sbux4['Calories'].mean(),
        dnkn4['Calories'].mean()
    ]
})

# Define brand colors
starbucks_green = '#00704A'
dunkin_pink = '#FF59E9'

# Create bar chart
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    avg_calories['Company'],
    avg_calories['Average_Calories'],
    color=[starbucks_green, dunkin_pink]
)

# Add value labels
ax.bar_label(bars, fmt='%.1f', padding=5, fontweight='bold')

# Add title and axis labels
ax.set_title('Average Beverage Calories',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Company')
ax.set_ylabel('Calories')

plt.tight_layout()
plt.show()
```

This chart compares the average calories in beverages from Starbucks and Dunkin' Donuts. Starbucks beverages have an average of 230.8 calories, while Dunkin' Donuts beverages average 253.3 calories. Although the difference is not very large, Dunkin' has slightly more calories per beverage on average. This suggests that Starbucks may offer somewhat lower-calorie options overall.

## Task 3.5: Comparing Healthier Beverage Choices

This chart compares how many beverages at Starbucks and Dunkin' Donuts have fewer than 100 calories and fewer than 15 grams of sugar.

```python
# Count beverages meeting both conditions
sbux_healthy = len(
    sbux4[
        (sbux4['Calories'] < 100) &
        (sbux4['Sugar_g'] < 15)
    ]
)

dnkn_healthy = len(
    dnkn4[
        (dnkn4['Calories'] < 100) &
        (dnkn4['Sugars (g)'] < 15)
    ]
)

# Create comparison dataframe
healthy = pd.DataFrame({
    'Company': ['Starbucks', "Dunkin' Donuts"],
    'Healthy_Beverages': [sbux_healthy, dnkn_healthy]
})

# Create bar chart
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    healthy['Company'],
    healthy['Healthy_Beverages'],
    color=[starbucks_green, dunkin_pink]
)

ax.bar_label(bars, padding=5, fontweight='bold')

ax.set_title('Healthier Beverage Portfolio',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Company')
ax.set_ylabel('Number of Beverages')

plt.tight_layout()
plt.show()
```

This chart compares the number of beverages that meet the healthier beverage criteria used in the analysis. Starbucks has 144 beverages that meet these requirements, while Dunkin' Donuts has 88. Although Starbucks has more healthier beverage options in total, it also has a larger overall beverage portfolio. This means that the number of healthier options should also be considered in relation to the total number of beverages offered by each company.

## Task 3.6: Competitive Positioning Map

This scatter plot compares the sugar and calorie content of beverages from Starbucks and Dunkin' Donuts. Each point represents one beverage combination.

```python
fig, ax = plt.subplots(figsize=(10, 7))

# Starbucks beverages
ax.scatter(
    sbux4['Sugar_g'],
    sbux4['Calories'],
    color=starbucks_green,
    alpha=0.35,
    label='Starbucks'
)

# Dunkin' Donuts beverages
ax.scatter(
    dnkn4['Sugars (g)'],
    dnkn4['Calories'],
    color=dunkin_pink,
    alpha=0.35,
    label="Dunkin' Donuts"
)

ax.set_title('Competitive Beverage Positioning Map',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Sugar (g)')
ax.set_ylabel('Calories')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

This scatter plot compares the sugar and calorie content of Starbucks and Dunkin' Donuts beverages. Both companies show a positive relationship between sugar and calories, meaning that drinks with more sugar generally contain more calories. Most beverages from both companies fall within similar ranges, but Dunkin' Donuts has more extreme values, including some drinks with over 1,000 calories. This suggests that Dunkin' offers some higher-calorie beverage options compared to Starbucks.

## Task 3.9: Top 5 Most Caffeinated Beverages
### Starbucks

```python
# Sort Starbucks beverages by caffeine content
top_caffeine_sbux = sbux4.sort_values(
    by='Caffeine_mg', ascending=False
).head(5).copy()

# Combine beverage name and size
top_caffeine_sbux['Label'] = (
    top_caffeine_sbux['Product_Name'] +
    ' (' + top_caffeine_sbux['Size'] + ')'
)

# Create horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 6))

bars = ax.barh(
    top_caffeine_sbux['Label'],
    top_caffeine_sbux['Caffeine_mg'],
    color='#00704A'
)

ax.bar_label(bars, padding=5)

ax.set_title('Top 5 Most Caffeinated Starbucks Beverages',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Caffeine (mg)')
ax.invert_yaxis()

plt.tight_layout()
plt.show()
```

The chart shows the five Starbucks beverages with the highest caffeine content. The True North Blend Blonde Roast in a Venti size has the most caffeine at 475 mg, followed by the Clover Medium Roast at 445 mg. Most of the beverages in the top five are Venti sizes, and all five are brewed coffees. This shows that Starbucks' larger brewed coffee options tend to have some of the highest caffeine amounts in the dataset.

Task 3.11: Average Fat Content by Milk Type

For this analysis, I want to compare the average total fat content of Starbucks beverages based on the type of milk used. I chose this because customers may want to know which milk options are associated with lower-fat drinks. The chart will help show how the average fat content differs across the milk options in the dataset.

```python
# Task 3.11: Average fat content by milk type

milk_fat = (
    sbux4.groupby('Milk')['Total_Fat_g']
    .mean()
    .sort_values()
)

fig, ax = plt.subplots(figsize=(10, 5))

bars = ax.barh(
    milk_fat.index,
    milk_fat.values,
    color='#00704A'
)

ax.bar_label(bars, fmt='%.1f', padding=4)

ax.set_title('Average Starbucks Beverage Fat Content by Milk Type',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Average Total Fat (g)')
ax.set_ylabel('Milk Type')

plt.tight_layout()
plt.show()
```

The chart compares the average total fat content of Starbucks beverages based on the type of milk used. Whole milk has the highest average at 10.5 grams, followed by coconut milk at 8.4 grams and 2% milk at 7.8 grams. Nonfat milk has a lower average of 3.4 grams, while beverages without milk have the lowest average at 0.4 grams. Based on these results, drinks made with nonfat milk tend to have less fat than those made with whole milk. However, other ingredients and drink sizes can also affect the total fat content.
